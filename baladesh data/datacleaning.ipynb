Code xử lỹ chuẩn hóa cấu trúc 

```python
def structural_cleaning(df):
    # Tạo bản sao
    df_clean = df.copy()

    # 1. Nhất quán văn bản (Cực kỳ quan trọng cho District và Crop Name)
    # Loại bỏ khoảng trắng thừa ở đầu/cuối và đưa về chữ thường để tránh trùng lặp ẩn
    text_cols = ['District', 'Crop Name', 'Season', 'Growth', 'Harvest', 'Transplant']
    for col in text_cols:
        if col in df_clean.columns:
            df_clean[col] = df_clean[col].astype(str).str.strip().str.lower()
    
    # Ví dụ: Xử lý thủ công nếu bạn phát hiện lỗi đánh máy cụ thể
    # mapping = {'ha noi': 'hanoi', 'hcm city': 'ho chi minh'}
    # df_clean['District'] = df_clean['District'].replace(mapping)

    # 2. Định dạng kiểu dữ liệu
    # Ép kiểu các cột chỉ số thực vật và thời tiết về float64
    float_cols = ['Avg Temp', 'Avg Humidity', 'Rainfall', 'pH', 'Organic_Carbon', 'Nitrogen']
    # Tìm tất cả các cột NDVI, EVI, LAI, FPAR, LST để ép kiểu float
    ndvi_related = [c for c in df_clean.columns if any(x in c for x in ['NDVI', 'EVI', 'LAI', 'FPAR', 'LST'])]
    
    for col in float_cols + ndvi_related:
        if col in df_clean.columns:
            df_clean[col] = pd.to_numeric(df_clean[col], errors='coerce')

    # Ép kiểu các cột định lượng lớn về int64
    int_cols = ['Production', 'Max Temp', 'Heat_Stress_Days']
    for col in int_cols:
        if col in df_clean.columns:
            # fillna(0) trước khi ép kiểu int vì cột chứa NaN không thể ép thành int
            df_clean[col] = pd.to_numeric(df_clean[col], errors='coerce').fillna(0).astype('int64')

    # 3. Xóa bỏ các cột thừa (Redundant)
    # 'grid_log' thường là ID nội bộ của hệ thống thu thập, không có giá trị dự báo
    # 'valid_points' nếu giá trị giống nhau cho mọi dòng cũng nên xóa
    redundant_cols = ['grid_log', 'valid_points']
    df_clean.drop(columns=[c for c in redundant_cols if c in df_clean.columns], inplace=True)

    return df_clean

# Thực thi
df = structural_cleaning(df)
print("--- Cấu trúc sau khi dọn dẹp ---")
print(df.info())
```

Xử lý những giá trị năng xuất không hợp lệ = 0 hoặc vô lý

```python

def clean_production_and_invalid(df):
    df_clean = df.copy()
    
    # 1. Chuyển đổi cột 'Production' sang số, ép các giá trị rác (chữ, ký tự lạ) thành NaN
    # errors='coerce' sẽ biến "không xác định" thành NaN
    df_clean['Production'] = pd.to_numeric(df_clean['Production'], errors='coerce')
    
    # 2. Xử lý giá trị 0
    # Trong Data Cleaning, ta coi 0 là giá trị không hợp lệ cho năng suất
    # Bạn có thể chọn biến 0 thành NaN để xóa đồng nhất ở bước sau
    df_clean['Production'] = df_clean['Production'].replace(0, np.nan)
    
    # 3. Xóa các dòng có Production là NaN (bao gồm: rác ban đầu, chữ ban đầu và số 0)
    initial_shape = df_clean.shape[0]
    df_clean = df_clean.dropna(subset=['Production'])
    
    print(f"Đã loại bỏ {initial_shape - df_clean.shape[0]} dòng có năng suất không hợp lệ (rác hoặc bằng 0).")
    
    # 4. Tự động tìm và xóa các cột có quá nhiều giá trị rác hoặc toàn 0
    # Nếu một cột có hơn 90% là 0 hoặc NaN, cột đó không có giá trị dự báo
    for col in df_clean.columns:
        if df_clean[col].isnull().mean() > 0.9 or (df_clean[col] == 0).mean() > 0.9:
            df_clean.drop(columns=[col], inplace=True)
            print(f"Đã xóa cột '{col}' do chứa quá nhiều dữ liệu rác/số 0.")
            
    return df_clean

# Thực thi
df = clean_production_and_invalid(df)
```

```python


def clean_features_only(df):
    df_clean = df.copy()
    
    # 1. Chỉ lọc Logic cho các biến môi trường (Bỏ qua Production)
    logic_conditions = (
        (df_clean['pH'].between(0, 14)) & 
        (df_clean['Avg Humidity'].between(0, 100)) &
        (df_clean['Rainfall'] >= 0)
    )
    
    # Lọc NDVI nếu có
    ndvi_cols = [c for c in df_clean.columns if 'NDVI' in c]
    for col in ndvi_cols:
        logic_conditions &= df_clean[col].between(-1, 1)
        
    df_clean = df_clean[logic_conditions]
    print(f"Đã dọn dẹp xong Logic cho các biến đầu vào. Giữ nguyên cột Production.")

    # 2. Xử lý Outliers cho Features (Ví dụ: Rainfall quá cao do bão)
    # Chúng ta chỉ xử lý Outliers cho các cột đầu vào để mô hình ổn định
    feature_cols_to_fix = ['Rainfall', 'Avg Temp', 'Avg Humidity']
    
    for col in feature_cols_to_fix:
        if col in df_clean.columns:
            Q1 = df_clean[col].quantile(0.25)
            Q3 = df_clean[col].quantile(0.75)
            IQR = Q3 - Q1
            # Thay vì xóa, ta có thể "Capping" (Giới hạn) để không mất dòng
            lower_limit = Q1 - 1.5 * IQR
            upper_limit = Q3 + 1.5 * IQR
            df_clean[col] = df_clean[col].clip(lower=lower_limit, upper=upper_limit)
            
    print("Đã thực hiện Capping Outliers cho các biến thời tiết để tránh nhiễu.")
    return df_clean

# Thực thi
df = clean_features_only(df)
```

```python
import pandas as pd
import numpy as np
from scipy.stats import skew

def handle_skewness(df):
    df_transformed = df.copy()
    
    # 1. Xác định các cột số (loại bỏ Production nếu bạn muốn giữ nguyên)
    num_cols = df_transformed.select_dtypes(include=[np.number]).columns.tolist()
    if 'Production' in num_cols:
        num_cols.remove('Production') 
    
    # 2. Tính toán độ lệch hiện tại
    skewed_feats = df_transformed[num_cols].apply(lambda x: skew(x.dropna())).sort_values(ascending=False)
    
    # 3. Lọc ra các cột có độ lệch tuyệt đối > 0.75
    high_skew = skewed_feats[abs(skewed_feats) > 0.75].index
    print(f"Các cột bị lệch nặng cần xử lý: {list(high_skew)}")
    
    # 4. Áp dụng Log Transformation (Log1p để xử lý cả giá trị 0)
    # Công thức: x' = log(1 + x)
    for col in high_skew:
        df_transformed[col] = np.log1p(df_transformed[col])
        
    print("\n--- Đã hoàn thành khử lệch bằng Log Transformation ---")
    # Kiểm tra lại độ lệch sau khi xử lý
    new_skew = df_transformed[high_skew].apply(lambda x: skew(x.dropna()))
    print("Độ lệch mới sau xử lý:")
    print(new_skew)
    
    return df_transformed

# Thực thi
df = handle_skewness(df)
```
