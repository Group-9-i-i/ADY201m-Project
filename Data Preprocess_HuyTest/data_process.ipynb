# Xóa cột AP Ratio

```python
import pandas as pd

# 1. Đọc file dữ liệu
file_path = 'Agri_Data_Cleaned.csv'
df = pd.read_csv(file_path)

# 2. Kiểm tra và xóa cột 'AP Ratio'
if 'AP Ratio' in df.columns:
    df = df.drop(columns=['AP Ratio'])
    print("Đã xóa cột 'AP Ratio'.")
else:
    print("Không tìm thấy cột 'AP Ratio' trong file.")

# 3. Lưu kết quả ra file mới (để giữ nguyên file gốc)
output_file = 'Agri_Data_Cleaned_No_AP_Ratio.csv'
df.to_csv(output_file, index=False)

print(f"File đã được lưu tại: {output_file}")

# Hiển thị 5 dòng đầu để kiểm tra
print(df.head())
```
