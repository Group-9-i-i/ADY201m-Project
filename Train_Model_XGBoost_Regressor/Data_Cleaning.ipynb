# CLEAR COLUMNS NOT NEEDED

```python
import pandas as pd

# 1. Đọc dữ liệu
df = pd.read_csv('Bangladesh_database_Final_Merged.csv')
print("--- BẮT ĐẦU DỌN DẸP DỮ LIỆU ---")
print("1. Số lượng hàng và cột ban đầu:", df.shape)

# 2. Xóa các hàng trùng lặp (Duplicate rows)
df = df.drop_duplicates()
print("2. Số lượng hàng sau khi xóa các hàng trùng lặp:", len(df))

# 3. Xóa các CỘT không cần thiết và không có giá trị cho hồi quy
columns_to_drop = ['grid_log', 'valid_points', 'Year', 'Compaction_Risk']
df_cleaned = df.drop(columns=columns_to_drop)

# 4. Lọc bỏ các HÀNG trống / thiếu dữ liệu
# Xóa các hàng trống hoàn toàn (tất cả các cột đều không có dữ liệu)
df_cleaned = df_cleaned.dropna(how='all')
# Xóa các hàng có chứa ít nhất 1 ô trống (NaN)
df_cleaned = df_cleaned.dropna()
print("3. Số lượng hàng sau khi xóa hàng trống và hàng chứa NaN:", len(df_cleaned))

# 5. Xóa các hàng theo điều kiện Logic
# Xóa các hàng có Area = 0
df_cleaned = df_cleaned[df_cleaned['Area'] != 0]

# Xóa các hàng chứa lỗi '#DIV/0!' ở cột 'AP Ratio'
df_cleaned = df_cleaned[df_cleaned['AP Ratio'] != '#DIV/0!']
print("4. Số lượng hàng sau khi xóa Area=0 và lỗi #DIV/0!:", len(df_cleaned))

# 6. Ép kiểu dữ liệu cho hồi quy
# Do dính chữ '#DIV/0!' nên cột 'AP Ratio' đang bị hiểu là Object (Text). 
# Phải ép nó về dạng số (Float) thì hồi quy mới chạy được.
df_cleaned['AP Ratio'] = pd.to_numeric(df_cleaned['AP Ratio'], errors='coerce')

# Dropna một lần nữa phòng trường hợp hàm to_numeric tạo ra NaN ở các ô text lạ khác
df_cleaned = df_cleaned.dropna()

# >>> THÊM MỚI: TÍNH CỘT YIELD <<<
# Năng suất (Yield) = Sản lượng (Production) / Diện tích (Area)
df_cleaned['Yield'] = df_cleaned['Production'] / df_cleaned['Area']

# 7. Kiểm tra lại kết quả cuối cùng
print("5. Số lượng hàng và cột CUỐI CÙNG (Sẵn sàng hồi quy):", df_cleaned.shape)

# 8. Lưu lại thành file mới
df_cleaned.to_csv('Bangladesh_database_Cleaned.csv', index=False)
print("--- ĐÃ LƯU FILE THÀNH CÔNG ---")

# Kiểm tra lại file sạch
df_final = pd.read_csv('Bangladesh_database_Cleaned.csv')
display(df_final.head()) # Dùng display() nếu ở trong Jupyter Notebook, hoặc dùng print(df_final.head())
```

# Chuẩn hóa các cột số

```python
import pandas as pd

# 1. Đọc dữ liệu từ file
df = pd.read_csv('Bangladesh_database_Cleaned.csv')

# 2. Lấy danh sách tất cả các cột mang kiểu dữ liệu số (int, float)
numeric_cols = df.select_dtypes(include=['number']).columns

# 3. Làm tròn về 2 chữ số sau dấu phẩy cho các cột này
df[numeric_cols] = df[numeric_cols].round(2)

# 4. Xuất ra một file CSV mới
output_filename = 'Bangladesh_database_Rounded.csv'
df.to_csv(output_filename, index=False)

print(f"Đã lưu file thành công: {output_filename}")
```

# Xử lý các biến phân loại (Categorical Encoding) và Tách file 80% Train 20% test

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split

# 1. Đọc dữ liệu và lọc lỗi
df = pd.read_csv('Bangladesh_database_Rounded.csv')
df = df[(df['Production'] > 0) & (df['Area'] > 0)].copy()

# =====================================================================
# BƯỚC 1: BIẾN ĐỔI LOGARIT VÀ XỬ LÝ BIẾN ORDINAL
# =====================================================================
# Log Transform cho biến mục tiêu
df['Log_Yield'] = np.log1p(df['Yield'])
df.drop(columns=['Yield'], inplace=True)

# Map các biến Ordinal
ph_map = {'Acidic (Chua)': 0, 'Optimal (Tối ưu)': 1}
water_map = {'Moderate': 0, 'Optimal': 1}
heat_map = {'Low Risk': 0, 'Moderate Risk': 1, 'High Risk': 2}
texture_map = {'Clayey (Sét)': 0, 'Loamy (Thịt/Phù sa)': 1}

df['pH_Suitability'] = df['pH_Suitability'].map(ph_map)
df['Water_Availability_Cat'] = df['Water_Availability_Cat'].map(water_map)
df['Extreme_Heat_Risk'] = df['Extreme_Heat_Risk'].map(heat_map)
df['Dominant_Soil_Texture'] = df['Dominant_Soil_Texture'].map(texture_map)

# =====================================================================
# BƯỚC 2: CHIA TẬP DỮ LIỆU (TRAIN 80% - TEST 20%) THEO TỪNG CROP NAME
# =====================================================================
# Tham số stratify=df['Crop Name'] đảm bảo lấy đúng 20% của từng cây trồng
train_df, test_df = train_test_split(
    df, 
    test_size=0.2, 
    stratify=df['Crop Name'], 
    random_state=42 # Cố định random để chạy lại kết quả không đổi
)

print(f"Số lượng dòng tập Train: {len(train_df)}")
print(f"Số lượng dòng tập Test: {len(test_df)}")

# =====================================================================
# BƯỚC 3: TARGET ENCODING (CHỈ HỌC TỪ TẬP TRAIN)
# =====================================================================
nominal_cols = ['District', 'Season', 'Crop Name', 'Transplant', 'Growth', 'Harvest']

# Tính trung bình Log_Yield của toàn bộ tập Train (dùng để điền khuyết nếu cần)
global_mean_yield = train_df['Log_Yield'].mean()

for col in nominal_cols:
    # 1. TÍNH trung bình Log_Yield cho từng nhãn CHỈ TRÊN TẬP TRAIN
    target_mean = train_df.groupby(col)['Log_Yield'].mean()
    
    # 2. ÁP DỤNG (map) giá trị đó vào tập Train
    train_df[col] = train_df[col].map(target_mean)
    
    # 3. ÁP DỤNG (map) cùng giá trị đó vào tập Test
    test_df[col] = test_df[col].map(target_mean)
    
    # 4. Xử lý ngoại lệ: Nếu tập Test xuất hiện một nhãn (ví dụ District mới) 
    # mà tập Train không hề có, nó sẽ bị NaN. Ta điền bằng global_mean_yield
    test_df[col].fillna(global_mean_yield, inplace=True)

# =====================================================================
# BƯỚC 4: XUẤT FILE ĐỂ TRAIN MÔ HÌNH
# =====================================================================
train_df.to_csv('Train_Data_XGBoost.csv', index=False)
test_df.to_csv('Test_Data_XGBoost.csv', index=False)

print("Đã tách và xử lý xong! Bạn sẽ có 2 file: Train_Data_XGBoost.csv và Test_Data_XGBoost.csv")
```

# Xác định biến mục tiêu (Target) và biến độc lập (Features)

```python
import pandas as pd

train_df = pd.read_csv('Train_Data_XGBoost.csv')
test_df = pd.read_csv('Test_Data_XGBoost.csv')

# Loại bỏ 'Log_Yield' (Target), 'Production' và 'AP Ratio' (Cột gây rò rỉ dữ liệu)
drop_cols = ['Log_Yield', 'Production', 'AP Ratio']

X_train = train_df.drop(columns=drop_cols)
y_train = train_df['Log_Yield']

X_test = test_df.drop(columns=drop_cols)
y_test = test_df['Log_Yield']

# Lưu lại 4 file CSV mới
X_train.to_csv('X_train.csv', index=False)
y_train.to_csv('y_train.csv', index=False)
X_test.to_csv('X_test.csv', index=False)
y_test.to_csv('y_test.csv', index=False)

print("Đã loại bỏ AP Ratio và tạo lại 4 file thành công!")
```
